```python
import subprocess, sys

def run_benchmark(name, folder):
    datapath = DATA_ROOT / folder
    if not datapath.is_dir():
        raise FileNotFoundError(f"Dataset not found: {datapath}")
    subprocess.run([
        sys.executable, "main.py", "--benchmark", name,
        "--datapath", str(datapath), "--nshot", "1",
        "--adapt-to", "first-episode"
    ], check=True)
```

```python
run_benchmark("isic", "isic2018-classwise")
```

```python
run_benchmark("lung", "lungsegmentation")
```

```python
run_benchmark("deepglobe", "deepglobe")
```

```python
run_benchmark("suim", "suim")
```

```python
run_benchmark("fss", "fss1000")
```
